### Q4 
Ek file `numbers.txt` banayein aur us mein 1 se 5 tak numbers line-by-line write karein.

```python
# Q4 solution code here
with open('number.txt','w') as file:
    for i in range(1, 6):
        data = file.write(str(i) + "\n")
        print(i)
```

### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
# Q5 solution code here
total_sum = 0

with open("number.txt", "r") as f:
    for line in f:
        total_sum += int(line.strip())
        
        print("Numbers ka total sum hai:", total_sum)
    
```

### Q6

```python
# Q6 solution code here
with open("test.txt", "r") as file:
    lines = file.readlines()
    total_lines = len(lines)
    print(f" {total_lines} lines hain.")
```
